```python
def trans_features(df):
    new_df = df.copy()
    new_df["Street"] = df["Street"].map({"Grvl": 0, "Pave": 1}).astype("int64")
    new_df["Alley"] = df["Alley"].map({"Grvl": 1, "Pave": 2}).fillna(0).astype("int64")
    new_df["LotShape"] = df["LotShape"].map({"Reg": 3, "IR1": 2, "IR2": 1, "IR3": 0}).astype("int64")
    new_df["LandContour"] = df["LandContour"].map({"Lvl": 3, "Bnk": 2, "HLS": 1, "Low": 0}).astype("int64")
    new_df["Utilities"] = df["Utilities"].map({"AllPub": 3, "NoSewr": 2, "NoSeWa": 1, "ELO": 0}).astype("int64")
    new_df["LandSlope"] = df["LandSlope"].map({"Gtl": 2, "Mod": 1, "Sev": 0}).astype("int64")
    new_df["ExterQual"] = df["ExterQual"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).astype("int64")
    new_df["ExterCond"] = df["ExterCond"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).astype("int64")
    new_df["BsmtQual"] = df["BsmtQual"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).fillna(0).astype("int64")
    new_df["BsmtCond"] = df["BsmtCond"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).fillna(0).astype("int64")
    new_df["BsmtExposure"] = df["BsmtExposure"].map({"Gd": 4, "Av": 3, "Mn": 2, "No": 1}).fillna(0).astype("int64")
    new_df["BsmtFinType1"] = df["BsmtFinType1"].map({"GLQ": 6, "ALQ": 5, "BLQ": 4, "Rec": 3, "LwQ": 2, "Unf": 1}).fillna(0).astype("int64")
    new_df["BsmtFinType2"] = df["BsmtFinType2"].map({"GLQ": 6, "ALQ": 5, "BLQ": 4, "Rec": 3, "LwQ": 2, "Unf": 1}).fillna(0).astype("int64")
    new_df["HeatingQC"] = df["HeatingQC"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).astype("int64")
    new_df["CentralAir"] = df["CentralAir"].map({"N": 0, "Y": 1}).astype("int64")
    new_df["KitchenQual"] = df["KitchenQual"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).astype("int64")
    new_df["Functional"] = df["Functional"].map({"Typ": 8, "Min1": 7, "Min2": 6, "Mod": 5, "Maj1": 4, "Maj2": 3, "Sev": 2, "Sal": 1}).astype("int64")
    new_df["FireplaceQu"] = df["FireplaceQu"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).fillna(0).astype("int64")
    new_df["GarageQual"] = df["GarageQual"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).fillna(0).astype("int64")
    new_df["GarageCond"] = df["GarageCond"].map({"Ex": 5, "Gd": 4, "TA": 3, "Fa": 2, "Po": 1}).fillna(0).astype("int64")
    new_df["PavedDrive"] = df["PavedDrive"].map({"Y": 2, "P": 1, "N": 0}).astype("int64")
    new_df["PoolQC"] = df["PoolQC"].map({"Ex": 4, "Gd": 3, "TA": 2, "Fa": 1,}).fillna(0).astype("int64")

    new_df["LotFrontage"] = df["LotFrontage"].fillna(0)
    new_df["MasVnrArea"] = df["MasVnrArea"].fillna(0)
    new_df["GarageYrBlt"] = df["GarageYrBlt"].fillna(0)

    return new_df
```

```python
def normalize_features(train_df, curr_df):
    curr_df["LotFrontage"] = (curr_df["LotFrontage"] - train_df["LotFrontage"].mean()) / train_df["LotFrontage"].std()
    curr_df["LotArea"] = (curr_df["LotArea"] - train_df["LotArea"].mean()) / train_df["LotArea"].std()
    curr_df["YearBuilt"] = (curr_df["YearBuilt"] - train_df["YearBuilt"].mean()) / train_df["YearBuilt"].std()
    curr_df["YearRemodAdd"] = (curr_df["YearRemodAdd"] - train_df["YearRemodAdd"].mean()) / train_df["YearRemodAdd"].std()
    curr_df["MasVnrArea"] = (curr_df["MasVnrArea"] - train_df["MasVnrArea"].mean()) / train_df["MasVnrArea"].std()
    curr_df["BsmtFinSF1"] = (curr_df["BsmtFinSF1"] - train_df["BsmtFinSF1"].mean()) / train_df["BsmtFinSF1"].std()
    curr_df["BsmtFinSF2"] = (curr_df["BsmtFinSF2"] - train_df["BsmtFinSF2"].mean()) / train_df["BsmtFinSF2"].std()
    curr_df["BsmtUnfSF"] = (curr_df["BsmtUnfSF"] - train_df["BsmtUnfSF"].mean()) / train_df["BsmtUnfSF"].std()
    curr_df["TotalBsmtSF"] = (curr_df["TotalBsmtSF"] - train_df["TotalBsmtSF"].mean()) / train_df["TotalBsmtSF"].std()
    curr_df["1stFlrSF"] = (curr_df["1stFlrSF"] - train_df["1stFlrSF"].mean()) / train_df["1stFlrSF"].std()
    curr_df["2ndFlrSF"] = (curr_df["2ndFlrSF"] - train_df["2ndFlrSF"].mean()) / train_df["2ndFlrSF"].std()
    curr_df["LowQualFinSF"] = (curr_df["LowQualFinSF"] - train_df["LowQualFinSF"].mean()) / train_df["LowQualFinSF"].std()
    curr_df["GrLivArea"] = (curr_df["GrLivArea"] - train_df["GrLivArea"].mean()) / train_df["GrLivArea"].std()
    curr_df["BsmtFullBath"] = (curr_df["BsmtFullBath"] - train_df["BsmtFullBath"].mean()) / train_df["BsmtFullBath"].std()
    curr_df["BsmtHalfBath"] = (curr_df["BsmtHalfBath"] - train_df["BsmtHalfBath"].mean()) / train_df["BsmtHalfBath"].std()
    curr_df["FullBath"] = (curr_df["FullBath"] - train_df["FullBath"].mean()) / train_df["FullBath"].std()
    curr_df["HalfBath"] = (curr_df["HalfBath"] - train_df["HalfBath"].mean()) / train_df["HalfBath"].std()
    curr_df["BedroomAbvGr"] = (curr_df["BedroomAbvGr"] - train_df["BedroomAbvGr"].mean()) / train_df["BedroomAbvGr"].std()
    curr_df["KitchenAbvGr"] = (curr_df["KitchenAbvGr"] - train_df["KitchenAbvGr"].mean()) / train_df["KitchenAbvGr"].std()
    curr_df["TotRmsAbvGrd"] = (curr_df["TotRmsAbvGrd"] - train_df["TotRmsAbvGrd"].mean()) / train_df["TotRmsAbvGrd"].std()
    curr_df["Fireplaces"] = (curr_df["Fireplaces"] - train_df["Fireplaces"].mean()) / train_df["Fireplaces"].std()
    curr_df["GarageYrBlt"] = (curr_df["GarageYrBlt"] - train_df["GarageYrBlt"].mean()) / train_df["GarageYrBlt"].std()
    curr_df["GarageCars"] = (curr_df["GarageCars"] - train_df["GarageCars"].mean()) / train_df["GarageCars"].std()
    curr_df["GarageArea"] = (curr_df["GarageArea"] - train_df["GarageArea"].mean()) / train_df["GarageArea"].std()
    curr_df["WoodDeckSF"] = (curr_df["WoodDeckSF"] - train_df["WoodDeckSF"].mean()) / train_df["WoodDeckSF"].std()
    curr_df["OpenPorchSF"] = (curr_df["OpenPorchSF"] - train_df["OpenPorchSF"].mean()) / train_df["OpenPorchSF"].std()
    curr_df["EnclosedPorch"] = (curr_df["EnclosedPorch"] - train_df["EnclosedPorch"].mean()) / train_df["EnclosedPorch"].std()
    curr_df["3SsnPorch"] = (curr_df["3SsnPorch"] - train_df["3SsnPorch"].mean()) / train_df["3SsnPorch"].std()
    curr_df["ScreenPorch"] = (curr_df["ScreenPorch"] - train_df["ScreenPorch"].mean()) / train_df["ScreenPorch"].std()
    curr_df["PoolArea"] = (curr_df["PoolArea"] - train_df["PoolArea"].mean()) / train_df["PoolArea"].std()
    curr_df["MiscVal"] = (curr_df["MiscVal"] - train_df["MiscVal"].mean()) / train_df["MiscVal"].std()

    return curr_df
```

```python
oh_columns = [
    "MSSubClass", "MSZoning",
    "LotConfig", "Neighborhood", "Condition1", "Condition2", "BldgType",
    "HouseStyle",  "RoofStyle", "RoofMatl", "Exterior1st", "Exterior2nd", "MasVnrType",
    "Foundation", "Heating", "Electrical", "GarageType", "GarageFinish", "Fence", "MiscFeature",
    "SaleType", "SaleCondition"
]
# asis_columns = [
#     "LotFrontage", "LotArea", "OverallQual", "OverallCond", "YearRemodAdd", "MasVnrArea",
#     "BsmtFinSF1", "BsmtFinSF2", "BsmtUnfSF", "TotalBsmtSF", "1stFlrSF", "2ndFlrSF",
#     "LowQualFinSF", "GrLivArea", "BsmtFullBath", "BsmtHalfBath", "FullBath", "HalfBath",
#     "Bedroom", "Kitchen", "TotRmsAbvGrd", "Fireplaces", "GarageCars",
#     "GarageArea", "WoodDeckSF", "OpenPorchSF", "EnclosedPorch", "3SsnPorch", "ScreenPorch",
#     "PoolArea", "MiscVal", "GarageYrBlt"
# ]
# already_processed = ["Street", "CentralAir", "PavedDrive"]
# dropped_columns = ["YearBuilt", "MoSold", "YrSold", "SaleType", "SaleCondition", "GarageYrBlt"]
dropped_columns = []
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

def encode_features(train_df, curr_df, oh_columns):
    encoder.fit_transform(train_df[oh_columns]),
    encoded_data = pd.DataFrame(
        encoder.transform(curr_df[oh_columns]),
        columns=encoder.get_feature_names_out(oh_columns),
        index=curr_df.index
    )
    curr_df.drop(columns=oh_columns, inplace=True)
    curr_df = pd.concat([curr_df, encoded_data], axis=1)
    return curr_df
```

```python
X_train = trans_features(X_train)
X_train = normalize_features(X_train, X_train)
X_train = encode_features(X_train, X_train, oh_columns)

X_test = trans_features(X_test)
X_test = normalize_features(X_train, X_test)
X_test = encode_features(X_test, X_test, oh_columns)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(
    n_estimators=500,
    max_depth=50,
    random_state=42
)
```

```python
from sklearn.metrics import root_mean_squared_error

rf.fit(X_train, y_train)
training_predictions = rf.predict(X_train)
root_mean_squared_error(y_true=y_train, y_pred=training_predictions)
```

```python
testing_split_predictions = rf.predict(X_test)
root_mean_squared_error(y_true=y_test, y_pred=testing_split_predictions)
```
