# Training data

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier
import optuna
import torch
import numpy as np
```

```python
train_data = pd.read_csv("digit-recognizer/train.csv")
train_data
```

```python
train_data["pixel300"].unique()
```

```python
feature_set = train_data.drop(columns="label")
target_set = train_data["label"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    feature_set,
    target_set,
    random_state=42,
    shuffle=True,
    train_size=0.7
)
```

# Test data

```python
test_data = pd.read_csv("digit-recognizer/test.csv")
test_data
```

# Models

#### RF

```python
rf = RandomForestClassifier(
    max_depth=50,
    n_estimators=400,
    random_state=42
)

rf.fit(X_train, y_train)
```

```python
predictions = rf.predict(X_train)
accuracy_score(y_pred=predictions, y_true=y_train)
```

```python
predictions = rf.predict(X_test)
accuracy_score(y_pred=predictions, y_true=y_test)
```

#### XGB

```python
xgb = XGBClassifier(
    max_depth=50,
    n_estimators=10,
    num_parallel_tree=10,
    random_state=42
)

xgb.fit(X_train, y_train)
```

```python
predictions = xgb.predict(X_train)
accuracy_score(y_pred=predictions, y_true=y_train)
```

```python
predictions = xgb.predict(X_test)
accuracy_score(y_pred=predictions, y_true=y_test)
```

#### Neural Net

```python
torch.manual_seed(0)
```

```python
class NeuralNetwork(torch.nn.Module):
    def __init__(self, input_units, output_units):
        super().__init__()

        # self.layers = torch.nn.Sequential(
        #     torch.nn.Linear(input_units, 256),
        #     torch.nn.ReLU(),
        #     torch.nn.Linear(256, 128),
        #     # torch.nn.ReLU(),
        #     # torch.nn.Linear(98, 28),
        #     torch.nn.ReLU(),
        #     torch.nn.Linear(128, output_units)
        # )
        self.layers = torch.nn.Sequential(
            torch.nn.Linear(input_units, 1024),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.3),
            torch.nn.Linear(1024, 2048),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.3),
            # torch.nn.Linear(2048, 1024),
            # torch.nn.ReLU(),
            # torch.nn.Dropout(0.3),
            torch.nn.Linear(2048, output_units)
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits
```

```python
model = NeuralNetwork(784, 10)
model
```

```python
numpy_feature_set = feature_set.to_numpy().astype(np.float32)
numpy_target_set = target_set.to_numpy().astype(np.long)

numpy_X_train, numpy_X_test, numpy_y_train, numpy_y_test = train_test_split(
    numpy_feature_set,
    numpy_target_set,
    random_state=42,
    shuffle=True,
    train_size=0.7
)
```

```python
with torch.no_grad():
    out = torch.argmax(model(torch.from_numpy(numpy_X_test)), dim=1)
print(out)
```

```python
class NNDataset(torch.utils.data.Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]

train_ds = NNDataset(numpy_X_train, numpy_y_train)
test_ds = NNDataset(numpy_X_test, numpy_y_test)

train_loader = torch.utils.data.DataLoader(
    dataset=train_ds,
    batch_size=32,
    shuffle=True,
    num_workers=0,
    drop_last=True
)

test_loader = torch.utils.data.DataLoader(
    dataset=test_ds,
    batch_size=32,
    shuffle=True,
    num_workers=2
)
```

```python
import torch.nn.functional as F

optimizer = torch.optim.SGD(model.parameters(), lr=0.001, weight_decay=1e-3)

epochs = 15

for epoch in range(epochs):
    model.train()

    for batch_idx, (features, labels) in enumerate(train_loader):
        logits = model(features)
        loss = F.cross_entropy(logits, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        print(f"Epoch {epoch+1:03d}/{epochs:03d} | Batch {batch_idx:03d}/{len(train_loader):03d} | Train loss: {loss:.2f}")

    model.eval()
```

#### CNN

```python
import torch
import numpy as np

torch.manual_seed(0)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class ConvNeuralNetwork(torch.nn.Module):
    def __init__(self, input_channels, output_units):
        super().__init__()

        # self.layers = torch.nn.Sequential(
        #     torch.nn.Linear(input_units, 256),
        #     torch.nn.ReLU(),
        #     torch.nn.Linear(256, 128),
        #     # torch.nn.ReLU(),
        #     # torch.nn.Linear(98, 28),
        #     torch.nn.ReLU(),
        #     torch.nn.Linear(128, output_units)
        # )
        self.layers = torch.nn.Sequential(
            torch.nn.Conv2d(in_channels=input_channels, out_channels=8, kernel_size=3, padding=1),
            torch.nn.MaxPool2d(kernel_size=2, stride=2),
            torch.nn.Conv2d(in_channels=8, out_channels=16, kernel_size=3, padding=1),
            torch.nn.MaxPool2d(kernel_size=2, stride=2),
            torch.nn.Conv2d(in_channels=16, out_channels=32, kernel_size=3, padding=1),
            torch.nn.MaxPool2d(kernel_size=2, stride=2, ceil_mode=True),
            torch.nn.Flatten(),
            torch.nn.Linear(32 * 4 * 4, 784),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.2),
            torch.nn.Linear(784, 256),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.2),
            torch.nn.Linear(256, 128),
            torch.nn.ReLU(),
            torch.nn.Dropout(0.2),
            # torch.nn.Linear(2048, 1024),
            # torch.nn.ReLU(),
            # torch.nn.Dropout(0.3),
            torch.nn.Linear(128, output_units)
        )

    def forward(self, x):
        logits = self.layers(x)
        return logits


model = ConvNeuralNetwork(1, 10)
# model = NeuralNetwork(4, 2)
model

numpy_feature_set = feature_set.to_numpy().astype(np.float32)
numpy_target_set = target_set.to_numpy().astype(np.long)
numpy_feature_set = numpy_feature_set.reshape(-1, 1, 28, 28)

numpy_X_train, numpy_X_test, numpy_y_train, numpy_y_test = train_test_split(
    numpy_feature_set,
    numpy_target_set,
    random_state=42,
    shuffle=True,
    train_size=0.7
)


class NNDataset(torch.utils.data.Dataset):
    def __init__(self, X, y):
        self.features = X
        self.labels = y

    def __getitem__(self, index):
        one_x = self.features[index]
        one_y = self.labels[index]
        return one_x, one_y

    def __len__(self):
        return self.labels.shape[0]

train_ds = NNDataset(numpy_X_train, numpy_y_train)
test_ds = NNDataset(numpy_X_test, numpy_y_test)

train_loader = torch.utils.data.DataLoader(
    dataset=train_ds,
    batch_size=32,
    shuffle=True,
    num_workers=0,
    drop_last=True
)

test_loader = torch.utils.data.DataLoader(
    dataset=test_ds,
    batch_size=32,
    shuffle=True,
    num_workers=2
)

model.to(device)


import torch.nn.functional as F

# optimizer = torch.optim.SGD(model.parameters(), lr=0.0001, weight_decay=1e-4)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-4)

epochs = 15

for epoch in range(epochs):
    model.train()

    for batch_idx, (features, labels) in enumerate(train_loader):
        features = features.to(device)
        labels = labels.to(device)
        logits = model(features)
        loss = F.cross_entropy(logits, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        print(f"Epoch {epoch+1:03d}/{epochs:03d} | Batch {batch_idx:03d}/{len(train_loader):03d} | Train loss: {loss:.2f}")

    model.eval()
```
